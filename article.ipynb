As the Globalise project reaches its conclusion in 2026, this contribution reflects on the development of a comprehensive research environment designed to make an important part of the Dutch East India Company (VOC) archives more accessible and researchable. Funded by the Dutch Research Council (NWO), Globalise focuses on the "Received Letters and Papers": an archival series of over five million handwritten pages held at the Dutch National Archives in The Hague, which represents a colonially mediated record of events and processes in the Indian Ocean and Indonesian archipelago in the seventeenth- and eighteenth-century.


While these materials are now fully scanned and globally accessible, a major issue persists. Traditional archival indexing has overwhelmingly centered on Western actors and the VOC's institutional history of maritime trade and profit. This focus has severely marginalized the visibility of non-Western individuals and societies within these sources.


To address this imbalance, Globalise adopts a more inclusive, data-driven approach. The project began by compiling extensive reference data from a wide range of secondary literature, forming a comprehensive knowledge base that encompasses people, polities, places, organizations, ships, commodities, professions, social statuses and more. These data were used to identify and contextualize entities recognized within the documents through Named Entity Recognition (NER). The resulting entity profiles were further enriched by deploying a specialized model to detect and classify the historical events in which these actors played a role. By systematically incorporating a broad spectrum of historical actors, the project aims to offer a more representative view of a corpus inherently shaped by the perspectives of the VOC and its clerks.


A significant gap in current digital history infrastructures is the limited focus on end-users, particularly those in the Global South. Globalise addresses this through an infrastructure explicitly designed to be plurivocal: rather than reproducing colonial taxonomies, it provides contextual layers for archival concepts in multiple languages, including contested terms, with multi-layered provenance information and interpretative framing. With this approach, the colonial archives are accessible not only to specialists but to anyone with an interest in the histories this archive tells. It also provides users with the necessary historical context to devise new and diverse histories.


Recognizing that computational methods are not neutral, the project foregrounds transparency at all stages of data creation and modeling. This includes detailed documentation of the reference data's provenance, the contributions of external collaborators, and the identification and mitigation of algorithmic biases. To ensure broad and equitable access, Globalise makes its infrastructure and datasets fully open, adhering strictly to FAIR principles and established best practices in digital heritage and the humanities. This maximizes the accessibility, interoperability, and reusability of its outputs, significantly enhancing their research value.


Globalise provides an infrastructure that prioritizes 21st-century contextualization over the uncritical reading of colonial archives and replication of colonial taxonomies. With the project's unique integration of archival textual data with carefully curated contextual information, and transparency in used computational approaches, users can proficiently navigate what was previously a closed (for experts only) colonial archive, opening it up across linguistic boundaries, cultural backgrounds and research capabilities. We thereby offer a model for how digital history can contribute to the decolonization of archival knowledge.
